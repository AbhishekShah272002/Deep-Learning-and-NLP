```python
import torch

print(torch.__version__)
print(torch.cuda.is_available())
```

```python
# Creating Cpu example
first_tens = torch.tensor([1, 2, 3])
second_tens = torch.tensor([4,5,6])
print(first_tens)
print(second_tens)
add_tens = first_tens + second_tens
print(add_tens)
```

```python
sub_tens = first_tens - second_tens
print(sub_tens)
print(sub_tens.shape)
```

```python
# Creating Gpu example
if torch.cuda.is_available(): device = "cuda"
else: device = "cpu"
print(device)
```

```python
tens_a = torch.tensor([1,2,3]).to(device)
tens_b = torch.tensor([4,5,6]).to(device)
add_tens = tens_a + tens_b
print(add_tens)
```
