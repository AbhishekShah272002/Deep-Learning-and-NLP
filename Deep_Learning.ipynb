```python
import torch

print(torch.__version__)
print(torch.cuda.is_available())
```

```python
# Creating Cpu example
first_tens = torch.tensor([1, 2, 3])
second_tens = torch.tensor([4,5,6])
print(first_tens)
print(second_tens)
add_tens = first_tens + second_tens
print(add_tens)
```

```python
sub_tens = first_tens - second_tens
print(sub_tens)
print(sub_tens.shape)
```

```python
# Creating Gpu example
if torch.cuda.is_available(): device = "cuda"
else: device = "cpu"
print(device)
```

```python
tens_a = torch.tensor([1,2,3]).to(device)
tens_b = torch.tensor([4,5,6]).to(device)
add_tens = tens_a + tens_b
print(add_tens)
```

```python
# Using torch.empty(), torch.ones(), and torch
empty_tens = torch.empty(2,3)
zero_tens = torch.zeros(2,3)
ones_tens = torch.ones(2,3)
print("empty_tens: ",empty_tens)
print("zero_tens ",zero_tens)
print("ones_tens ",ones_tens)
```

```python
# Filled with random numbers from a uniform distribution
rand_un = torch.rand(2,2)
print(rand_un)
# Filled with random numbers from a normal distribution
rand_norm = torch.randn(2,2)
print(rand_norm)
# Filled with random integers generated uniformaly
rand_int = torch.randint(10,100,(2,2))
print(rand_int)
```
