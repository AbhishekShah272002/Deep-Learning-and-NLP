```python
import pandas as pd
import re
import nltk

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
```

```python
df = pd.read_csv(
    "ecommerce_customer_feedback.txt",
    header=None,
    names=["Customer_Feedback"],
)
df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
df["Cleaned_Feedback"] = df["Customer_Feedback"].str.lower()

df[["Customer_Feedback", "Cleaned_Feedback"]].head()
```

```python
df["Cleaned_Feedback"] = df["Cleaned_Feedback"].apply(
    lambda text: re.sub(r'[^a-z\s]', '', text)
)

df[["Customer_Feedback", "Cleaned_Feedback"]].head()
```
