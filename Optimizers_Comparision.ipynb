```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import time

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import SGD, Adam, RMSprop
```

```python
df = pd.read_csv("breastcancer.csv")

df.head()
```

```python
x = df.drop(['diagnosis', 'id'], axis=1)

y = df['diagnosis']
y = y.map({'M':0, 'B':1})

print("x shape: ", x.shape)
print("y shape: ", y.shape)
print(y.value_counts())
```

```python
y = y.map({'M':0, 'B':1})

print(y.value_counts())
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

y_train = y_train.to_numpy()
y_test = y_test.to_numpy()

print("Training Date: ", x_train.shape)
print("Testing Date: ", x_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)

print("Standardization Completed ")
```

```python
x_train
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

def create_model():
    model = Sequential()

    model.add(Dense(16, activation='relu', input_dim=30))
    model.add(Dense(8, activation='relu'))
    model.add(Dense(1, activation='sigmoid'))

    return model
```

```python
from tensorflow.keras.optimizers import SGD

# Create a fresh ANN model
sgd_model = create_model()

# Compile using SGD
sgd_model.compile(
    optimizer=SGD(learning_rate=0.01),
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# Train the model
history_sgd = sgd_model.fit(
    x_train,
    y_train,
    epochs=30,
    batch_size=32,
    validation_split=0.2,
    verbose=1
)
```

```python
from tensorflow.keras.optimizers import Adam

# Create a fresh ANN model
adam_model = create_model()

# Compile using Adam
adam_model.compile(
    optimizer=Adam(learning_rate=0.001),
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# Train the model
history_adam = adam_model.fit(
    x_train,
    y_train,
    epochs=30,
    batch_size=32,
    validation_split=0.2,
    verbose=1
)
```

```python
from tensorflow.keras.optimizers import RMSprop

# Create a fresh ANN model
rmsprop_model = create_model()

# Compile using RMSProp
rmsprop_model.compile(
    optimizer=RMSprop(learning_rate=0.001),
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# Train the model
history_rmsprop = rmsprop_model.fit(
    x_train,
    y_train,
    epochs=30,
    batch_size=32,
    validation_split=0.2,
    verbose=1
)
```

```python
sgd_loss, sgd_accuracy = sgd_model.evaluate(x_test, y_test, verbose=0)

adam_loss, adam_accuracy = adam_model.evaluate(x_test, y_test, verbose=0)

rmsprop_loss, rmsprop_accuracy = rmsprop_model.evaluate(x_test, y_test, verbose=0)

print("SGD Test Accuracy     :", sgd_accuracy)
print("SGD Test Loss         :", sgd_loss)

print("\nAdam Test Accuracy    :", adam_accuracy)
print("Adam Test Loss        :", adam_loss)

print("\nRMSProp Test Accuracy :", rmsprop_accuracy)
print("RMSProp Test Loss     :", rmsprop_loss)
```

```python
plt.figure(figsize=(8,5))

plt.plot(history_sgd.history['loss'], label='SGD')
plt.plot(history_adam.history['loss'], label='Adam')
plt.plot(history_rmsprop.history['loss'], label='RMSProp')

plt.xlabel('Epoch')
plt.ylabel('Training Loss')
plt.title('Loss Curve Comparison of Optimizers')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.plot(history_sgd.history['accuracy'], label='SGD')
plt.plot(history_adam.history['accuracy'], label='Adam')
plt.plot(history_rmsprop.history['accuracy'], label='RMSProp')

plt.xlabel('Epoch')
plt.ylabel('Training Accuracy')
plt.title('Accuracy Comparison of Optimizers')
plt.legend()
plt.show()
```

```python
results = pd.DataFrame({
    'Optimizer': ['SGD', 'Adam', 'RMSProp'],
    'Test Accuracy': [sgd_accuracy, adam_accuracy, rmsprop_accuracy],
    'Test Loss': [sgd_loss, adam_loss, rmsprop_loss]
})

results
```
