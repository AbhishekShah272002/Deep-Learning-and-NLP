```python
from sklearn.preprocessing import LabelEncoder

# Initialize the LabelEncoder
label_encoder = LabelEncoder()

# Apply Label Encoding to the 'diagnosis' column
df['diagnosis_encoded'] = label_encoder.fit_transform(df['diagnosis'])

# Display the first few rows with the new encoded column
display(df[['diagnosis', 'diagnosis_encoded']].head())
```

```python
# Display the mapping of original categories to encoded numbers
print("Original categories:", label_encoder.classes_)
print("Encoded numbers:", label_encoder.transform(label_encoder.classes_))
```

### Normalization
Normalization is a technique often applied as part of data preparation for machine learning. The goal of normalization is to change the values of numeric columns in the dataset to a common scale, without distorting differences in the ranges of values. This is especially important for algorithms that are sensitive to the magnitudes of input features, such as K-Nearest Neighbors, Support Vector Machines, or neural networks.

One common method is **Min-Max Normalization** (or scaling), which scales features to a fixed range, usually 0 to 1. The formula for Min-Max Normalization for a value `x` is:

`x_normalized = (x - min(x)) / (max(x) - min(x))`

Where `min(x)` and `max(x)` are the minimum and maximum values of the feature, respectively.

```python
from sklearn.preprocessing import MinMaxScaler

# Identify numerical columns to normalize
# Exclude 'id' as it's just an identifier and 'diagnosis'/'diagnosis_encoded' as they are target/encoded categorical.
numerical_cols = df.select_dtypes(include=np.number).columns.tolist()
numerical_cols.remove('id')
numerical_cols.remove('diagnosis_encoded')

print(f"Columns to be normalized: {numerical_cols}")

# Initialize the MinMaxScaler
scaler = MinMaxScaler()

# Apply Min-Max Scaling to the selected numerical columns
df_normalized = df.copy() # Create a copy to store normalized data
df_normalized[numerical_cols] = scaler.fit_transform(df_normalized[numerical_cols])

# Display the first few rows of the normalized DataFrame
display(df_normalized.head())
```

### Model Training and Evaluation
Now that our data is preprocessed (Label Encoded and Normalized), we can proceed with training a classification model and evaluating its performance using various metrics.

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report

# Define features (X) and target (y)
# Drop 'id', 'diagnosis', and the original 'diagnosis_encoded' if not needed as a feature.
X = df_normalized.drop(columns=['id', 'diagnosis', 'diagnosis_encoded'])
y = df_normalized['diagnosis_encoded']

# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"y_test shape: {y_test.shape}")
```

```python
# Initialize and train a Logistic Regression model
model = LogisticRegression(random_state=42, solver='liblinear') # 'liblinear' is good for small datasets
model.fit(X_train, y_train)

# Make predictions on the test set
y_pred = model.predict(X_test)

# Display the first few predictions and true values
print("First 10 predictions:", y_pred[:10])
print("First 10 true values:", y_test.values[:10])
```

### Evaluation Metrics
Let's calculate and display the confusion matrix, accuracy score, and a detailed classification report.

```python
# Calculate the Confusion Matrix
conf_matrix = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:")
display(pd.DataFrame(conf_matrix,
                     index=['Actual Benign (0)', 'Actual Malignant (1)'],
                     columns=['Predicted Benign (0)', 'Predicted Malignant (1)']))

# Calculate Accuracy
accuracy = accuracy_score(y_test, y_pred)
print(f"\nAccuracy: {accuracy:.4f}")

# Generate a Classification Report
class_report = classification_report(y_test, y_pred, target_names=['Benign (0)', 'Malignant (1)'])
print("\nClassification Report:")
print(class_report)
```
